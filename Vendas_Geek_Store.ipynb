```python
import pandas as pd
```

```python
df_vendas_geek_store = pd.read_csv("/content/vendas_geek_store.csv")
display(df_vendas_geek_store)
```

```python
df_vendas_geek_store.describe()
```

```python
df_vendas_geek_store["receita"] = df_vendas_geek_store['Quantidade_Vendida'] * df_vendas_geek_store['Preco_Venda']
display(df_vendas_geek_store[["receita","Produto"]])
```

```python
df_vendas_geek_store["lucro unidade"] = df_vendas_geek_store["Preco_Venda"] - df_vendas_geek_store['Custo_Fabricante']
display(df_vendas_geek_store[["lucro unidade","Produto"]])
```

```python
df_vendas_geek_store["lucro final"] = df_vendas_geek_store["lucro unidade"] * df_vendas_geek_store['Quantidade_Vendida']
display(df_vendas_geek_store[["lucro final","Produto"]])
```

```python
display(df_vendas_geek_store[["Produto","Categoria"]])
```
